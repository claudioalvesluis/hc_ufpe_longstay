```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

# 1) Escolha o modelo
model_id = "Qwen/Qwen2.5-1.5B-Instruct"  # ou "mistralai/Mistral-7B-Instruct-v0.3"

# 2) Verifica GPU disponível
device = 0 if torch.cuda.is_available() else -1
print("Usando GPU" if device == 0 else "Sem GPU disponível")

# 3) Carrega modelo e tokenizador
tok = AutoTokenizer.from_pretrained(model_id)
mdl = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",         # distribui automaticamente entre GPUs (se houver mais de uma)
    torch_dtype=torch.float16  # reduz uso de memória e acelera inferência
)

# 4) Cria o pipeline com device configurado
gen = pipeline(
    "text-generation",
    model=mdl,
    tokenizer=tok,
    device=device
)

# 5) Teste
prompt = "Resuma o papel da inteligência artificial na medicina."
out = gen(prompt, max_new_tokens=80, temperature=0.7)
print(out[0]["generated_text"])
```
